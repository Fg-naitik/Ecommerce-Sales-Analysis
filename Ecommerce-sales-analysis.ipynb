```python
clean_df.isnull().sum()
```

```python
clean_df.groupby("Category")["Stock"].sum().sort_values(ascending=False).plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Total Stock by Category")
plt.xlabel("Category")
plt.ylabel("Total Stock")
plt.xticks(rotation=45)
plt.show()
```

```python
clean_df["Color"].value_counts().plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Number of Products by Color")
plt.xlabel("Color")
plt.ylabel("Number of Products")
plt.xticks(rotation=45)
plt.show()
```

```python
clean_df["Size"].value_counts().plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Number of Products by Size")
plt.xlabel("Size")
plt.ylabel("Number of Products")
plt.xticks(rotation=0)
plt.show()
```

```python
clean_df["Size"].value_counts().plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Number of Products by Size")
plt.xlabel("Size")
plt.ylabel("Number of Products")
plt.xticks(rotation=0)
plt.show()
```

```python
clean_df.groupby("Category")["Stock"].sum().sort_values(ascending=False).plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Total Stock by Category")
plt.xlabel("Category")
plt.ylabel("Total Stock")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.hist(clean_df["Stock"], bins=20)

plt.title("Distribution of Stock")
plt.xlabel("Stock")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# PIE CHART - Product Distribution by Size

size_counts = clean_df["Size"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    size_counts,
    labels=size_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Product Distribution by Size")
plt.show()
```

```python
# BOX PLOT - Stock Distribution

plt.figure(figsize=(8, 5))

sns.boxplot(x=clean_df["Stock"])

plt.title("Stock Distribution")
plt.xlabel("Stock")

plt.tight_layout()
plt.show()
```

```python
# HEATMAP - Category vs Size

category_size = pd.crosstab(
    clean_df["Category"],
    clean_df["Size"]
)

plt.figure(figsize=(12, 8))

sns.heatmap(
    category_size,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Product Count by Category and Size")
plt.xlabel("Size")
plt.ylabel("Category")

plt.tight_layout()
plt.show()
```

```python
# PIE CHART - Top 5 Categories

top_categories = clean_df["Category"].value_counts().head(5)

plt.figure(figsize=(8, 8))

plt.pie(
    top_categories,
    labels=top_categories.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Top 5 Categories by Product Count")
plt.show()
```

```python
# STACKED BAR CHART - Category and Size

category_size = pd.crosstab(
    clean_df["Category"],
    clean_df["Size"]
)

category_size.plot(
    kind="bar",
    stacked=True,
    figsize=(14, 7)
)

plt.title("Product Distribution by Category and Size")
plt.xlabel("Category")
plt.ylabel("Number of Products")
plt.xticks(rotation=45)
plt.legend(title="Size")

plt.tight_layout()
plt.show()
```

```python
# HORIZONTAL BAR CHART - Total Stock by Category

category_stock = (
    clean_df.groupby("Category")["Stock"]
    .sum()
    .sort_values()
)

category_stock.plot(
    kind="barh",
    figsize=(10, 7)
)

plt.title("Total Stock by Category")
plt.xlabel("Total Stock")
plt.ylabel("Category")

plt.tight_layout()
plt.show()
```

```python
# COUNT PLOT - Number of Products in Each Category

plt.figure(figsize=(12, 6))

sns.countplot(
    data=clean_df,
    x="Category",
    order=clean_df["Category"].value_counts().index
)

plt.title("Number of Products by Category")
plt.xlabel("Category")
plt.ylabel("Number of Products")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# BOX PLOT - Stock Distribution by Category

plt.figure(figsize=(14, 7))

sns.boxplot(
    data=clean_df,
    x="Category",
    y="Stock"
)

plt.title("Stock Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Stock")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
category_size = pd.crosstab(
    clean_df["Category"],
    clean_df["Size"]
)

plt.figure(figsize=(12, 8))

sns.heatmap(
    category_size,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.title("Product Count by Category and Size")
plt.xlabel("Size")
plt.ylabel("Category")

plt.tight_layout()
plt.show()
```

```python
category_counts = clean_df["Category"].value_counts()

plt.figure(figsize=(9, 9))

plt.pie(
    category_counts,
    labels=category_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Product Distribution by Category")
plt.show()
```

```python
category_stock = (
    clean_df.groupby("Category")["Stock"]
    .sum()
    .sort_values()
)

plt.figure(figsize=(10, 7))

category_stock.plot(kind="barh")

plt.title("Total Stock by Category")
plt.xlabel("Total Stock")
plt.ylabel("Category")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 7))

sns.violinplot(
    data=clean_df,
    x="Category",
    y="Stock"
)

plt.title("Stock Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Stock")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 6))

plt.plot(clean_df["Stock"])

plt.title("Stock Variation Across Products")
plt.xlabel("Product Index")
plt.ylabel("Stock")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.scatter(
    range(len(clean_df)),
    clean_df["Stock"],
    alpha=0.5
)

plt.title("Stock Distribution Across Products")
plt.xlabel("Product Number")
plt.ylabel("Stock")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.hist(
    clean_df["Stock"].dropna(),
    bins=30
)

plt.title("Distribution of Stock Levels")
plt.xlabel("Stock")
plt.ylabel("Number of Products")

plt.tight_layout()
plt.show()
```

```python
category_counts = clean_df["Category"].value_counts()

print("Top 10 Categories:")
print(category_counts.head(10))
```

```python
category_stock = (
    clean_df.groupby("Category")["Stock"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 Categories by Total Stock:")
print(category_stock.head(10))
```

```python
# ==========================================
# E-COMMERCE DATA ANALYSIS
# ==========================================

# 1. Which categories have the most products?
print("1. TOP CATEGORIES BY NUMBER OF PRODUCTS")
print(clean_df["Category"].value_counts().head(10))


# 2. Which categories have the most total stock?
print("\n2. TOP CATEGORIES BY TOTAL STOCK")
category_stock = (
    clean_df.groupby("Category")["Stock"]
    .sum()
    .sort_values(ascending=False)
)
print(category_stock.head(10))


# 3. Which sizes are most common?
print("\n3. MOST COMMON SIZES")
print(clean_df["Size"].value_counts().head(10))


# 4. Which colors are most common?
print("\n4. MOST COMMON COLORS")
print(clean_df["Color"].value_counts().head(10))


# 5. Which designs have the highest stock?
print("\n5. TOP DESIGNS BY STOCK")
design_stock = (
    clean_df.groupby("Design No.")["Stock"]
    .sum()
    .sort_values(ascending=False)
)
print(design_stock.head(10))


# 6. What is the average stock?
print("\n6. AVERAGE STOCK")
print(clean_df["Stock"].mean())


# 7. What is the highest stock?
print("\n7. HIGHEST STOCK")
print(clean_df["Stock"].max())


# 8. What is the lowest stock?
print("\n8. LOWEST STOCK")
print(clean_df["Stock"].min())


# 9. How many unique products/SKUs are there?
print("\n9. UNIQUE SKU CODES")
print(clean_df["SKU Code"].nunique())


# 10. How many missing values remain?
print("\n10. REMAINING MISSING VALUES")
print(clean_df.isnull().sum())


# 11. How many duplicate rows are there?
print("\n11. DUPLICATE ROWS")
print(clean_df.duplicated().sum())
```
